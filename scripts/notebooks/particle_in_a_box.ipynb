First, let's import the necessary libraries: `numpy` for numerical operations and `matplotlib.pyplot` for plotting. Then, we'll define the parameters for the particle in a box problem, such as the length of the box and physical constants.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define constants (using natural units for simplicity, or set to 1)
hbar = 1.0  # Reduced Planck's constant
m = 1.0     # Mass of the particle
L = 1.0     # Length of the box (from 0 to L)

# Number of eigenstates to visualize
num_states = 5

# Create a spatial grid for plotting the wave functions
x = np.linspace(0, L, 500) # 500 points from 0 to L
```

Now, we'll calculate the eigenvalues (energy levels) and eigenfunctions (wave functions) for the first `num_states` of the particle in a box. The formulas are:

*   **Eigenvalues (Energy):** $E_n = \frac{n^2 \pi^2 \hbar^2}{2mL^2}$
*   **Eigenfunctions (Wave function):** $\psi_n(x) = \sqrt{\frac{2}{L}} \sin\left(\frac{n\pi x}{L}\right)$

```python
energies = []
wave_functions = []

for n in range(1, num_states + 1):
    # Calculate eigenvalue (energy)
    En = (n**2 * np.pi**2 * hbar**2) / (2 * m * L**2)
    energies.append(En)

    # Calculate eigenfunction (wave function)
    psi_n = np.sqrt(2 / L) * np.sin(n * np.pi * x / L)
    wave_functions.append(psi_n)

# Print the calculated energies
print("Calculated Eigenvalues (Energy Levels):")
for i, E in enumerate(energies):
    print(f"E_{i+1} = {E:.4f}")
```

Finally, let's visualize these eigenfunctions and their corresponding energy levels. Each wave function will be plotted, and a horizontal line will indicate its energy level.

```python
plt.figure(figsize=(10, 15))
plt.rcParams.update({'font.size': 16}) # Global font size increase

# Set font family preferences to prioritize Times New Roman
plt.rcParams['font.serif'] = ['Times New Roman', 'Times', 'DejaVu Serif', 'serif']
plt.rcParams['font.family'] = 'serif'
plt.rcParams['mathtext.fontset'] = 'cm' # Use Computer Modern for math text, which is serif

# Factor to expand amplitudes for better visualization
amplitude_scaling_factor = 4.0

# Store handles and labels for a combined legend
energy_line_handles = []
energy_line_labels = []
psi_line_handles = []
psi_line_labels = []
psi_squared_line_handles = []
psi_squared_line_labels = []
node_handle = None # To store handle for node legend entry

for i in range(num_states):
    n = i + 1
    En = energies[i]
    psi_n = wave_functions[i]

    # Apply an additional offset for psi_5 and its square
    y_offset_psi = 0
    y_offset_psi_sq = 0
    if n == 5:
        y_offset_psi = 4 # Adjusted value for '0.5 inch lower'
        y_offset_psi_sq = 4 # Adjusted value for '0.5 inch lower'

    if n == 4:
        y_offset_psi = 3
        y_offset_psi_sq = 3

    if n == 3:
        y_offset_psi = 2
        y_offset_psi_sq = 2

    if n == 1:
        y_offset_psi = -1.5
        y_offset_psi_sq = -1.5


    # Scale the wave function amplitude
    psi_n_scaled = psi_n * amplitude_scaling_factor

    # Plot the energy level and store handle/label for the legend
    energy_line = plt.axhline(y=En, color='gray', linestyle='--', linewidth=0.8)
    energy_line_handles.append(energy_line)
    energy_line_labels.append(f'$E_{n}$ = {En:.2f}')

    # Add energy value text next to the line, now showing the full formula (far left side)
    plt.text(-0.01, En, rf'${n**2/2} \frac{{ \pi^2 \hbar^2}}{{ma^2}}$', verticalalignment='center', horizontalalignment='right', color='black', fontsize=14) # Specific font size for formula

    # Add psi_n label (left side, slightly to the right of the energy formula) with white background and black border
    plt.text(L * 0.02, En + 5 + y_offset_psi, rf'$\psi_{n}(x)$', verticalalignment='center', horizontalalignment='left', color='black', fontsize=14, bbox=dict(boxstyle="round,pad=0.1", fc="white", ec="black", lw=0.8, alpha=0.6))

    # Add psi_squared label (further right side) with black font and cyan background and black border
    plt.text(L * 0.91, En + 6.5 + y_offset_psi_sq, rf'$|\psi_{n}(x)|^2$', verticalalignment='center', horizontalalignment='left', color='black', fontsize=14, bbox=dict(boxstyle="round,pad=0.1", fc="cyan", ec="black", lw=0.8, alpha=0.6))

    # Create an invisible plot for the overall wave function label in the legend
    # This ensures each psi_n has a single entry in the legend, regardless of its colors
    psi_invisible_plot, = plt.plot(x, np.full_like(x, np.nan), color='black', linewidth=2) # Use an invisible line
    psi_line_handles.append(psi_invisible_plot)
    # Fixed: Using a raw string literal with concatenation for LaTeX labels to avoid MathText parsing errors
    psi_line_labels.append(r'$\psi_{%d}(x)$' % n)

    # Separate positive and negative parts of the scaled wave function
    # And add the energy offset for plotting
    psi_n_positive_vals = np.where(psi_n_scaled >= 0, psi_n_scaled + En, np.nan)
    psi_n_negative_vals = np.where(psi_n_scaled < 0, psi_n_scaled + En, np.nan)

    # Plot the positive part in red (without label as it's handled by the invisible plot)
    plt.plot(x, psi_n_positive_vals, color='red', linewidth=2)
    # Plot the negative part in blue (without label)
    plt.plot(x, psi_n_negative_vals, color='blue', linewidth=2)

    # Fill the area between the wave function and the energy level
    plt.fill_between(x, En, psi_n_positive_vals, color='red', alpha=0.2)
    plt.fill_between(x, En, psi_n_negative_vals, color='blue', alpha=0.2)

    # Calculate and plot the square of the wave function
    psi_n_squared = psi_n**2
    psi_n_scaled_squared = psi_n_squared * amplitude_scaling_factor # Use the same scaling factor
    psi_squared_plot, = plt.plot(x, psi_n_scaled_squared + En, color='cyan', linestyle='-', linewidth=1.5)
    psi_squared_line_handles.append(psi_squared_plot)
    psi_squared_line_labels.append(r'$|\psi_{%d}(x)|^2$' % n)

    # Add purple dots for nodes (where psi_n is zero, excluding boundaries)
    if n > 1: # The first state has no internal nodes
        node_positions = [k * L / n for k in range(1, n)] # Nodes at L/n, 2L/n, ..., (n-1)L/n
        # Plot nodes at the corresponding energy level
        node_plot, = plt.plot(node_positions, [En] * len(node_positions), 'o', color='purple', markersize=6)
        if node_handle is None: # Store handle for legend only once
            node_handle = node_plot

#plt.title('Eigenvalues and Eigenfunctions of a Particle in a Box', fontsize=16)
plt.xlabel('Position (x)', fontsize=16)
# Replaced plt.ylabel with plt.figtext for more control
plt.figtext(0.01, 0.5, 'Energy / Wave Function Amplitude', ha='left', va='center', fontsize=16, rotation=90) # Manual Y-axis label placement
plt.xlim(0, L)
plt.grid(True, linestyle=':', alpha=0.6)

# Customize x-axis tick labels to show '0' and 'a'
plt.xticks([0, L], ['0', '$a$'])
plt.yticks([]) # Keep y-axis numerical tick labels removed

# Create custom legend entries for the color coding explanation
from matplotlib.lines import Line2D
color_legend_handles = [Line2D([0], [0], color='red', lw=2),
                        Line2D([0], [0], color='blue', lw=2),
                        Line2D([0], [0], color='red', alpha=0.2, linewidth=8),
                        Line2D([0], [0], color='blue', alpha=0.2, linewidth=8),
                        Line2D([0], [0], color='cyan', lw=1.5)]
color_legend_labels = ['Positive Amplitude (Red)', 'Negative Amplitude (Blue)', 'Positive Amplitude Fill', 'Negative Amplitude Fill', r'Probability Density ($|\psi(x)|^2$)']

# Combine all handles and labels for a single legend
# Order: Energy levels, then Wave Functions, then Color Key, then Nodes
all_handles = energy_line_handles + psi_line_handles + psi_squared_line_handles + color_legend_handles
all_labels = energy_line_labels + psi_line_labels + psi_squared_line_labels + color_legend_labels

if node_handle: # Add node legend entry if nodes were plotted
    all_handles.append(node_handle)
    all_labels.append('Nodes')

#Remove legend as requested
# plt.legend(handles=all_handles, labels=all_labels, loc='lower center', bbox_to_anchor=(0.5, -0.25), ncol=3, fontsize='small')
plt.tight_layout() # Adjust layout without rect, allowing more space for labels

plt.show()
```

Now, let's visualize the square of the eigenfunctions, which represent the probability density functions ($|\psi_n(x)|^2$). The styling will be similar to the previous plot, including expanded amplitudes, '0' and 'a' labels for the x-axis, and the energy formulas on the right, but without the legend.

```python
plt.figure(figsize=(10, 15))
plt.rcParams.update({'font.size': 14}) # Global font size increase

# Factor to expand amplitudes for better visualization (can be adjusted for squared functions)
amplitude_scaling_factor = 4.0

# Store handles and labels for a combined legend (though legend will be removed)
energy_line_handles = []
energy_line_labels = []
psi_line_handles = []
psi_line_labels = []
node_handle = None # Nodes are still where psi is zero

for i in range(num_states):
    n = i + 1
    En = energies[i]
    psi_n = wave_functions[i]

    # Calculate the square of the wave function
    psi_n_squared = psi_n**2

    # Scale the squared wave function amplitude
    psi_n_scaled_squared = psi_n_squared * amplitude_scaling_factor

    # Plot the energy level
    energy_line = plt.axhline(y=En, color='gray', linestyle='--', linewidth=0.8)
    energy_line_handles.append(energy_line)
    energy_line_labels.append(f'$E_{n}$ = {En:.2f}')

    # Add energy value text next to the line, showing the full formula
    # Adjusted x-position (L * 1.02 instead of L * 1.05)
    plt.text(L * 1.02, En, rf'$|\psi_{n}(x)|^2$', verticalalignment='center', color='black', fontsize=12) # Specific font size for formula

    # Plot the squared wave function, shifted up by its energy
    # Since psi_n_squared is always positive, we only plot the 'positive' part in red.
    plt.plot(x, psi_n_scaled_squared + En, color='black', linewidth=2, label=rf'$|\psi_{n}(x)|^2$' if i == 0 else "")

    # Add purple dots for nodes (where psi_n is zero, excluding boundaries)
    if n > 1: # The first state has no internal nodes
        node_positions = [k * L / n for k in range(1, n)] # Nodes at L/n, 2L/n, ..., (n-1)L/n
        # Plot nodes at the corresponding energy level
        node_plot, = plt.plot(node_positions, [En] * len(node_positions), 'o', color='purple', markersize=6)
        if node_handle is None: # Store handle for legend only once
            node_handle = node_plot

#plt.title('Eigenvalues and Probability Density Functions ($|\psi(x)|^2$) of a Particle in a Box', fontsize=16)
plt.xlabel('Position (x)', fontsize=14)
plt.ylabel('Energy / Probability Density Amplitude', fontsize=14)
plt.xlim(0, L)
plt.grid(True, linestyle=':', alpha=0.6)

# Customize x-axis tick labels to show '0' and 'a'
plt.xticks([0, L], ['0', 'a'])
plt.yticks([]) # Keep y-axis numerical tick labels removed

# Combine all handles and labels for a single legend (though it will be commented out)
color_legend_handles = [Line2D([0], [0], color='red', lw=2)]
color_legend_labels = ['Probability Density Amplitude']

all_handles_for_legend = [handle for handle in psi_line_handles if handle] + color_legend_handles
all_labels_for_legend = [label for label in psi_line_labels if label] + color_legend_labels

if node_handle:
    all_handles_for_legend.append(node_handle)
    all_labels_for_legend.append('Nodes')

# Removed legend as requested in previous interactions
# plt.legend(handles=all_handles_for_legend, labels=all_labels_for_legend, loc='lower center', bbox_to_anchor=(0.5, -0.25), ncol=2, fontsize='small')
plt.tight_layout(rect=[0, 0, 1, 1]) # Adjust layout as legend is removed

plt.show()
```
